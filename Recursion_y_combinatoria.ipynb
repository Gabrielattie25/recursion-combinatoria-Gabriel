Factorial $n!$

````python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


for i in range(8):
    print(str(i) + "! =", factorial(i))
````

## Permutaciones

Una permutación es un arreglo ordenado de elementos. Dos permutaciones que
contienen los mismos elementos pero en distinto orden se consideran distintas.




```
FUNCIÓN dfs_permutaciones(lista):
    resultado ← []
    LLAMAR dfs([], lista, resultado)
    DEVOLVER resultado
FIN FUNCIÓN


PROCEDIMIENTO dfs(actual, restantes, resultado):
    // CASO BASE: no quedan elementos → permutación completa
    SI longitud(restantes) == 0 ENTONCES
        AGREGAR actual A resultado
        RETORNAR
    FIN SI

    PARA i DESDE 0 HASTA longitud(restantes) - 1 HACER
        siguiente    ← restantes[i]
        nuevo_resto  ← restantes SIN el elemento en posición i

        LLAMAR dfs(actual + [siguiente], nuevo_resto, resultado)
    FIN PARA
FIN PROCEDIMIENTO
```





````python
def dfs_permutaciones(lista):
    resultado = []
    dfs([], lista, resultado)
    return resultado


def dfs(actual, restantes, resultado):
    if len(restantes) == 0:
        resultado.append(actual)
        return

    for i in range(len(restantes)):
        siguiente = restantes[i]
        nuevo_resto = restantes[:i] + restantes[i + 1:]
        dfs(actual + [siguiente], nuevo_resto, resultado)


permutaciones = dfs_permutaciones([1, 2, 3])
for p in permutaciones:
    print(p)

print("Total de permutaciones:", len(permutaciones))
print("Comprobacion con 3!:", factorial(3))
````

```
FUNCIÓN dfs_combinaciones(lista, r):
    resultado ← []
    LLAMAR dfs([], 0, lista, r, resultado)
    DEVOLVER resultado
FIN FUNCIÓN


PROCEDIMIENTO dfs(actual, inicio, lista, r, resultado):
    // CASO BASE: ya elegimos r elementos
    SI longitud(actual) == r ENTONCES
        AGREGAR actual A resultado
        RETORNAR
    FIN SI

    // elegir el siguiente elemento SOLO desde `inicio` en adelante
    PARA i DESDE inicio HASTA longitud(lista) - 1 HACER
        LLAMAR dfs(actual + [lista[i]], i + 1, lista, r, resultado)
    FIN PARA
FIN PROCEDIMIENTO
```



````python
def dfs_combinaciones(lista, r):
    resultado = []
    dfs_comb([], 0, lista, r, resultado)
    return resultado


def dfs_comb(actual, inicio, lista, r, resultado):
    if len(actual) == r:
        resultado.append(actual)
        return

    for i in range(inicio, len(lista)):
        dfs_comb(actual + [lista[i]], i + 1, lista, r, resultado)


combinaciones = dfs_combinaciones([1, 2, 3, 4], 2)
for c in combinaciones:
    print(c)

n = 4
r = 2
print("Total de combinaciones:", len(combinaciones))
print("Comprobacion con n! / (r! * (n - r)!):", factorial(n) // (factorial(r) * factorial(n - r)))
````
